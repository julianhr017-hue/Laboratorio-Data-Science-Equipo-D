# Módulo 1: Extracción de Datos

## Ventas Regionales - Equipo D

En este módulo se procesarán los reportes de ventas correspondientes a tres sucursales: **Norte, Sur y Centro**.

El objetivo es automatizar la carga de los datos mediante un ciclo `for` y posteriormente unir los registros en un único DataFrame utilizando `pd.concat()`.

```python
import pandas as pd

# Sucursales asignadas al Equipo D
sucursales = ['norte', 'sur', 'centro']

print("Sucursales a procesar:")
for sucursal in sucursales:
    print("-", sucursal.capitalize())
```

```python
# URL pública del dataset
url = "https://raw.githubusercontent.com/selva86/datasets/master/supermarket_sales.csv"

# Cargar el archivo CSV desde la web
df_original = pd.read_csv(url)

# Relación entre las sucursales del dataset y las asignadas al Equipo D
mapa_sucursales = {
    'norte': 'A',
    'sur': 'B',
    'centro': 'C'
}

frames_equipo_d = []

# Recorrer las sucursales y extraer sus registros
for sucursal in sucursales:
    codigo = mapa_sucursales[sucursal]
    
    df_sucursal = df_original[df_original['Branch'] == codigo].copy()
    df_sucursal['Sucursal'] = sucursal.capitalize()
    
    frames_equipo_d.append(df_sucursal)

# Unir todas las sucursales en un solo DataFrame
df_ventas = pd.concat(frames_equipo_d, ignore_index=True)

print("Datos de ventas cargados correctamente.")
print("Dimensiones del DataFrame:", df_ventas.shape)
```

```python
# Mostrar las primeras filas del DataFrame final
print("Dimensiones del DataFrame final:", df_ventas.shape)

df_ventas.head()
```

```python
# Verificar la cantidad de registros por sucursal
conteo_sucursales = df_ventas['Sucursal'].value_counts()

print("Registros por sucursal:")
print(conteo_sucursales)
```

# Módulo 2: Manipulación y Limpieza

## Logística de Flotas - Equipo D

En este módulo se aplicará un filtro múltiple para seleccionar vehículos que cumplan simultáneamente las siguientes condiciones:

- Ser de la marca **Toyota o Honda**.
- Haber sido fabricados después de **2018**.
- Tener un kilometraje menor a **50,000 km**.

Para combinar las condiciones se utilizarán los operadores lógicos `|` (OR) y `&` (AND), agrupando cada condición mediante paréntesis.

```python
# Dataset de ejemplo para la misión del Equipo D
datos_autos = {
    'Marca': ['Toyota', 'Honda', 'Ford', 'Toyota', 'Honda',
              'Nissan', 'Toyota', 'Honda', 'Chevrolet', 'Toyota'],
    
    'Modelo': ['Corolla', 'Civic', 'Focus', 'RAV4', 'Accord',
               'Sentra', 'Camry', 'CR-V', 'Aveo', 'Yaris'],
    
    'Anio': [2020, 2021, 2019, 2017, 2022,
             2020, 2023, 2019, 2021, 2020],
    
    'Km': [35000, 42000, 25000, 30000, 52000,
           41000, 12000, 48000, 20000, 49000]
}

df_autos = pd.DataFrame(datos_autos)

print("Dataset original:")
df_autos
```

```python
# Filtro múltiple solicitado para el Equipo D
filtro = (
    ((df_autos['Marca'] == 'Toyota') | (df_autos['Marca'] == 'Honda'))
    & (df_autos['Anio'] > 2018)
    & (df_autos['Km'] < 50000)
)

df_filtrado_d = df_autos[filtro]

print("Vehículos que cumplen con todos los criterios:")
df_filtrado_d
```

# Módulo 3: Machine Learning

## Clasificación de Especies - Equipo D

En este módulo se utilizará el dataset `Iris` de Scikit-learn para entrenar un modelo de clasificación mediante **Support Vector Classifier (SVC)**.

Los datos se dividirán en conjuntos de entrenamiento y prueba, y posteriormente se utilizará `GridSearchCV` para encontrar la mejor combinación de hiperparámetros.

```python
# Importaciones para Machine Learning
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score

# Cargar el dataset Iris
X, y = load_iris(return_X_y=True)

# Dividir los datos en entrenamiento y prueba
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Datos cargados correctamente.")
print("Total de muestras:", X.shape[0])
print("Datos de entrenamiento:", X_train.shape)
print("Datos de prueba:", X_test.shape)
```

```python
# Crear el modelo SVC
modelo_svc = SVC()

# Hiperparámetros asignados al Equipo D
parametros = {
    'C': [0.5, 1, 5, 10],
    'kernel': ['linear', 'poly']
}

# Configurar GridSearchCV con validación cruzada
grid_search = GridSearchCV(
    estimator=modelo_svc,
    param_grid=parametros,
    cv=5
)

# Entrenar y probar todas las combinaciones
grid_search.fit(X_train, y_train)

# Mostrar los mejores resultados
print("Mejores parámetros:")
print(grid_search.best_params_)

print("\nMejor puntuación de validación cruzada:")
print(grid_search.best_score_)
```

```python
# Crear el modelo SVC
modelo_svc = SVC()

# Hiperparámetros asignados al Equipo D
parametros = {
    'C': [0.5, 1, 5, 10],
    'kernel': ['linear', 'poly']
}

# Configurar GridSearchCV
grid_search = GridSearchCV(
    estimator=modelo_svc,
    param_grid=parametros,
    cv=5
)

# Entrenar todas las combinaciones
grid_search.fit(X_train, y_train)

# Mostrar resultados
print("Mejores parámetros:")
print(grid_search.best_params_)

print("\nMejor puntuación de validación cruzada:")
print(grid_search.best_score_)
```

```python
# Obtener el mejor modelo encontrado por GridSearchCV
mejor_modelo = grid_search.best_estimator_

# Realizar predicciones con los datos de prueba
y_pred = mejor_modelo.predict(X_test)

# Calcular la precisión del modelo
precision = accuracy_score(y_test, y_pred)

print("Precisión del mejor modelo en los datos de prueba:")
print(precision)

print("\nPrecisión en porcentaje:")
print(f"{precision * 100:.2f}%")
```
